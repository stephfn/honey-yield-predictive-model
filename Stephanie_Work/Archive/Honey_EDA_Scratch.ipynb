```python
import os
import duckdb
from dotenv import load_dotenv

load_dotenv()

quack_host = os.getenv("TAILNET_HOST")
quack_token = os.getenv("QUACK_TOKEN")

con = duckdb.connect("honey.duckdb")

con.execute("FORCE INSTALL quack")
con.execute("LOAD quack")

try:
    con.execute("DETACH bees")
except:
    pass

con.execute(f"""
    ATTACH 'quack:{quack_host}' AS bees (
        TOKEN '{quack_token}',
        DISABLE_SSL true
    )
""")
```

```python
tables = con.execute("""
FROM bees.query('SHOW TABLES')
""").fetchdf()

tables
```

```python
# Inspect structure of the main sensor table

schema = con.execute("""
DESCRIBE bob_sensor_processed
""").fetchdf()

print(f"Number of columns: {len(schema)}")
schema
```
