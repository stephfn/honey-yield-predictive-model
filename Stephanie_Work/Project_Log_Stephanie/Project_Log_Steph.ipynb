### Key Findings
- Baseline model: MAE 0.768 kg, RMSE 2.437 kg, R² -0.296.
- Routine-only model: MAE 0.356 kg, RMSE 0.725 kg, R² 0.198.
- Routine hive changes are substantially more predictable than extreme events.
- Historical weight features and rolling averages provide the strongest predictive signal.
