```python
from dotenv import load_dotenv
import os
import duckdb

load_dotenv()

print("Host:", os.getenv("TAILNET_HOST"))

con = duckdb.connect()

con.execute(f"""
ATTACH 'quack:{os.getenv("TAILNET_HOST")}' AS bees (
    TOKEN '{os.getenv("QUACK_TOKEN")}',
    DISABLE_SSL true
)
""")

print("Connected!")
```
