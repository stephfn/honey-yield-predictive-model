# Executive Summary 

This part of the project established a robust data integration and predictive modeling pipeline to analyze hive weight changes. The primary objective was to move from manual observation to data-driven forecasting. While the initial modeling phase (Linear Regression, AdaBoost, and LightGBM) established a baseline performance, the project successfully identified critical "information ceilings" and variance drivers, providing a clear roadmap for future development.

# Methdology 

The modeling pipeline was developed using a time-series chronological split to prevent data leakage and ensure realistic forecasting. The project pipeline was constructed to allow for parallel experimentation, our group  implemented and benchmarked various machine learning architectures , mine are : Linear Regression, AdaBoost, and LightGBM to assess predictive viability across different algorithmic approaches.This  pipeline consists of three stages:

- Data Integration : Merging Historical Metrics unified THe SQL lite structure. 
- Feature Enginerring: Creation of Time based and rolling average to capture biological fluctuations 
- Modeling And Audit:  Benchmarking thre 3 machines learning architectures
